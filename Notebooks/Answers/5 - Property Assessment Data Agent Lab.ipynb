# Property assessment data agent lab

Create a Fabric data agent grounded in the Gold semantic model or Gold Lakehouse. Complete these steps in the Fabric portal and record your design decisions in new markdown cells.

## 1. Define the agent's job

Create an agent that answers questions about synthetic property values, comparable sales, inspections, appeal operations, and AI-derived appeal themes. Explicitly exclude real valuation advice, legal conclusions, and automated appeal decisions.

## 2. Select governed sources

Attach only the Gold semantic model or these curated marts:

- `property_assessment_mart`
- `comparable_sales_mart`
- `appeal_intelligence_mart`
- `neighborhood_equity_mart`

Do not ground the agent directly in raw narratives or Bronze data.

## 3. Add business instructions

Tell the agent that assessed value, sale price, and estimated tax are different measures. Define assessment-to-sale ratio, appeal rate, resolution rate, negative-sentiment rate, and average resolution days. Require the agent to state that all values are synthetic.

## 4. Add safety instructions

- Never claim that an assessment or tax calculation is legally authoritative.
- Never recommend approving or rejecting an appeal from sentiment.
- Distinguish source fields from AI-derived fields.
- State when the available data cannot answer a question.
- Ask for clarification when tax year, neighbourhood, or property class is ambiguous.

## 5. Evaluate the agent

Test at least these questions and inspect the generated query or grounding:

1. Which neighbourhood has the highest average assessed value for the latest tax year?
2. Which property classes have the largest assessment-to-sale ratio outliers?
3. What are the most common appeal reasons, and how often does the AI classification agree?
4. Which neighbourhoods have the highest negative appeal sentiment?
5. Show properties requiring inspection follow-up without treating that flag as a valuation decision.
6. Explain why an exact tax bill cannot be treated as legally authoritative in this dataset.

Compare numeric answers with the Gold tables. Refine agent instructions when the query uses the wrong grain, measure, or filter.

## Completion evidence

Record the final grounding source, agent instructions, evaluation questions, corrections made after testing, and one example where the agent correctly declined or qualified an unsupported request.
